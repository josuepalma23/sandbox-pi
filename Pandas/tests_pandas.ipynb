```python
## DATAFRAMES EN PANDAS
import pandas as pd

#   Un data frame es un formato tabular con filas y columnas para guardar datos

#   Usando un Diccionario + Lista
data = {"Nombre": ["Bob", "Patricio", "Calamardo"],
        "Edad": [30, 35, 50],
}

df = pd.DataFrame(data, index=["Empleado 1", "Empleado 2", "Empleado 3"])

#   acceder a una sola fila
#print(df.loc["Empleado 2"])

#   acceder por indice
#print(df.iloc[1]) #imprime Patricio

#   añadir una columna al dataFrame
#df["Posicion de trabajo"] = ["Cocinero", "Vago", "Cajero"]

#   añadir una fila al dataFrame (crear otro dataFrame) (siempre con mas diccionarios)
nuevas_fila = pd.DataFrame([{"Nombre": "Arenita", "Edad": 29, "Posicion de trabajo": "Ingeniera"},
                            {"Nombre": "Don Cangrejo", "Edad": 60, "Posicion de trabajo": "Jefe"}],
                           index=["Empleado 4", "Empleado 5"])

#   usamos concat, unimos los dataframes
df = pd.concat([df, nuevas_fila])

print(df)
```

```python
# FILTRADO EN PANDAS: obtener filas segun condiciones

import pandas as pd 

df = pd.read_csv("clientes_400.csv")


condicion = (df["satisfaccion"] > 3) & (df["numero_hijos"] > 0) & (df["ciudad"] == "Quito")

columns = ["cliente_id", "edad", "numero_hijos", "ciudad", "satisfaccion"]

nivel_alto_satisfaccion = df.loc[condicion, columns]

print(nivel_alto_satisfaccion.to_string())
```
