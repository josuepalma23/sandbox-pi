```python
import pandas as pd

clientes = pd.read_csv("clientes_400.csv")
clientes_muestra = clientes.sample(n=10, random_state=30)

# 1. Distribución porcentual en la Población y en la Muestra
distribucion_poblacion = clientes["edad"].value_counts(normalize=True) * 100
distribucion_muestra = clientes_muestra["edad"].value_counts(normalize=True) * 100

# 2. Crear el DataFrame
comparacion_clientes = pd.DataFrame({
    "Población (%)": distribucion_poblacion,
    "Muestra (%)": distribucion_muestra
}).fillna(0)

# 3. Calcular la diferencia
comparacion_clientes["Diferencia (%)"] = (
    comparacion_clientes["Muestra (%)"] - comparacion_clientes["Población (%)"]
)

# 4. CONVERTIR EL ÍNDICE 'EDAD' EN UNA COLUMNA NORMAL Y RENOMBRAR
comparacion_clientes = comparacion_clientes.reset_index()
comparacion_clientes.columns = ["Edad", "Población (%)", "Muestra (%)", "Diferencia (%)"]

# 5. Ordenar por edad para ver la tabla limpia
comparacion_clientes = comparacion_clientes.sort_values("Edad")

#print(comparacion_clientes.round(2).to_string(index=False))

# Filtrar para mostrar solo las edades presentes en la muestra de 10
muestra_filtrada = comparacion_clientes[comparacion_clientes["Muestra (%)"] > 0]

print(muestra_filtrada)
```

```python
# DEFINITIVO
import pandas as pd
from pandas.api.types import CategoricalDtype

clientes = pd.read_csv("clientes_400.csv")
clientes = clientes.dropna().drop_duplicates()

# siempre 2 decimales
pd.options.display.float_format = '{:.2f}'.format

#print(clientes)

#condicion = clientes[clientes["satisfaccion"] > 3]

#print(condicion)

# convertimos a categoricas y definimos prioridades 

tipo_educacion = CategoricalDtype(
    categories=["Primaria", "Secundaria", "Universidad", "Posgrado"],
    ordered=True
)

clientes["nivel_educacion"] = clientes["nivel_educacion"].astype(tipo_educacion)

nivel_satisfaccion = CategoricalDtype(
    categories=[1, 2, 3, 4, 5],
    ordered=True
)

clientes["satisfaccion"] = clientes["satisfaccion"].astype(nivel_satisfaccion)

#print(clientes.dtypes) ## podemos ver a ambos como categoria


## MAS (MUESTREO ALEATORIO SIMPLE)
muestra = 100

semilla = 42

muestra_simple = clientes.sample(n=muestra, random_state=semilla)

#print(muestra_simple.to_string())


## MUESTREO ESTRATIFICADO

fraccion = muestra / len(clientes)

muestra_estratificada = clientes.groupby("genero", group_keys=False).sample(
    frac=fraccion,
    random_state=semilla
)

#print(muestra_estratificada)
dist_poblacional = clientes["genero"].value_counts(normalize=True) * 100
dist_muestral = muestra_simple["genero"].value_counts(normalize=True) * 100
dist_estratificada = muestra_estratificada["genero"].value_counts(normalize=True) * 100

comparacion_entre_distribuciones = pd.DataFrame({
    "Poblacion (%)": dist_poblacional,
    "Muestreo Simple (%)": dist_muestral,
    "Muestreo Estratificado (%)": dist_estratificada
}).fillna(0)

comparacion_entre_distribuciones['Sesgo Simple (%)'] = (
    comparacion_entre_distribuciones['Muestreo Simple (%)'] - comparacion_entre_distribuciones['Poblacion (%)']
    )

comparacion_entre_distribuciones['Sesgo Estratificado (%)'] = (
    comparacion_entre_distribuciones['Muestreo Estratificado (%)'] - comparacion_entre_distribuciones['Poblacion (%)']
    )

print(comparacion_entre_distribuciones)
```

```python
import pandas as pd
from pandas.api.types import CategoricalDtype

# 1. Cargar y limpiar datos
clientes = pd.read_csv("clientes_400.csv")
clientes = clientes.dropna().drop_duplicates()

# Configuración de 2 decimales
pd.options.display.float_format = '{:.2f}'.format

# 2. Convertir a tipos categóricos ordinales
tipo_educacion = CategoricalDtype(
    categories=["Primaria", "Secundaria", "Universidad", "Posgrado"],
    ordered=True
)
clientes["nivel_educacion"] = clientes["nivel_educacion"].astype(tipo_educacion)

nivel_satisfaccion = CategoricalDtype(
    categories=[1, 2, 3, 4, 5],
    ordered=True
)
clientes["satisfaccion"] = clientes["satisfaccion"].astype(nivel_satisfaccion)

# 3. Parámetros de muestreo
muestra = 100
semilla = 42

# Muestreo Aleatorio Simple (MAS)
muestra_simple = clientes.sample(n=muestra, random_state=semilla)

# 4. MUESTREO ESTRATIFICADO POR 'nivel_educacion'
fraccion = muestra / len(clientes)

muestra_estratificada = clientes.groupby("nivel_educacion", group_keys=False).sample(
    frac=fraccion,
    random_state=semilla
)

# 5. Distribuciones por Nivel de Educación
dist_poblacional = clientes["nivel_educacion"].value_counts(normalize=True) * 100
dist_muestral = muestra_simple["nivel_educacion"].value_counts(normalize=True) * 100
dist_estratificada = muestra_estratificada["nivel_educacion"].value_counts(normalize=True) * 100

# 6. Tabla Comparativa
comparacion_educacion = pd.DataFrame({
    "Poblacion (%)": dist_poblacional,
    "Muestreo Simple (%)": dist_muestral,
    "Muestreo Estratificado (%)": dist_estratificada
}).fillna(0)

# .sort_index() asegura que se muestren en el orden de la jerarquía asignada
comparacion_educacion = comparacion_educacion.sort_index()

# 7. Cálculo de Sesgos
comparacion_educacion['Sesgo Simple (%)'] = (
    comparacion_educacion['Muestreo Simple (%)'] - comparacion_educacion['Poblacion (%)']
)

comparacion_educacion['Sesgo Estratificado (%)'] = (
    comparacion_educacion['Muestreo Estratificado (%)'] - comparacion_educacion['Poblacion (%)']
)

print("--- EVALUACIÓN DE SESGO EN NIVEL DE EDUCACIÓN ---")
print(comparacion_educacion)
```
